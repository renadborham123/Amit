```python
from pathlib import Path
from tempfile import TemporaryDirectory
from final_project import TextFileReader

with TemporaryDirectory() as directory:
    path = Path(directory) / 'sample.txt'
    path.write_text('Python makes data useful.\nOOP organizes code.', encoding='utf-8')
    reader = TextFileReader(path)
    reader.read_file()
    reader.display_content()
    print('Lines:', reader.count_lines())
    print('Words:', reader.count_words())
    print('Characters:', reader.count_characters())
```
